```python
def cost(board):
    n = len(board)
    attacks = 0

    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                attacks += 1

    return attacks

def print_board(board):
    for row in range(len(board)):
        print(" ".join("Q" if board[col] == row else "."
                       for col in range(len(board))))
    print()

def hill_climbing(board):
    while True:
        current_cost = cost(board)
        print(f"Current State: {board} | Cost: {current_cost}")
        print_board(board)

        if current_cost == 0:
            print("Goal reached!")
            return board

        best_board = board[:]
        best_cost = current_cost

        for col in range(len(board)):
            for row in range(len(board)):
                if row == board[col]:
                    continue

                neighbor = board[:]
                neighbor[col] = row
                neighbor_cost = cost(neighbor)

                if neighbor_cost < best_cost:
                    best_board = neighbor
                    best_cost = neighbor_cost

        if best_cost >= current_cost:
            print("Local optimum reached. Restart needed.")
            return None

        board = best_board

initial_board = [0, 0, 0, 0]
hill_climbing(initial_board)
```

```python


import random
import math


def cost(board):
    n = len(board)
    attacks = 0

    for i in range(n):
        for j in range(i + 1, n):
            if (board[i] == board[j] or
                    abs(board[i] - board[j]) == abs(i - j)):
                attacks += 1

    return attacks


def print_board(board):
    for row in range(8):
        print(" ".join(
            "Q" if board[col] == row else "."
            for col in range(8)
        ))
    print()


def simulated_annealing():
    board = [random.randint(0, 7) for _ in range(8)]
    current_cost = cost(board)

    temperature = 100.0
    cooling_rate = 0.99
    min_temperature = 0.01

    while temperature > min_temperature:
        if current_cost == 0:
            print("Solution found!")
            print("Board:", board)
            print_board(board)
            return

        neighbor = board[:]
        col = random.randint(0, 7)
        new_row = random.randint(0, 7)

        while new_row == neighbor[col]:
            new_row = random.randint(0, 7)

        neighbor[col] = new_row
        new_cost = cost(neighbor)

        delta = new_cost - current_cost

        if delta < 0 or random.random() < math.exp(
                -delta / temperature):
            board = neighbor
            current_cost = new_cost

        temperature *= cooling_rate

    if current_cost == 0:
        print("Solution found!")
        print("Board:", board)
        print_board(board)
    else:
        print("No solution found in this run.")
        print("Final cost:", current_cost)


simulated_annealing()
```
