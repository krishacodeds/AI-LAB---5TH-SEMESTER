```python
import itertools

def evaluate(expr, model):
    return eval(expr, {}, model)

def entails(kb, query, symbols):
    for values in itertools.product([True, False], repeat=len(symbols)):
        model = dict(zip(symbols, values))
        kb_true = all(evaluate(expr, model) for expr in kb)
        if kb_true and not evaluate(query, model):
            return False
    return True

if __name__ == "__main__":
    kb = [
        "(P == Q)",
        "(Q or R)",
        "(P or R)"
    ]
    symbols = ["P", "Q", "R"]


    queries = {
        "Q": "Q",
        "P->R": "(not P) or R",
        "R->Q": "(not R) or Q"
    }

    for name, expr in queries.items():
        result = entails(kb, expr, symbols)
        print(f"Does KB entail {name}? {result}")
```
