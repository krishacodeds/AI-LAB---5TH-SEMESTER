```python
board = [1, 2, 3, 4, 5, 6, 7, 8, 9]
player = "X"

for turn in range(9):
    print(board[0], board[1], board[2])
    print(board[3], board[4], board[5])
    print(board[6], board[7], board[8])

    move = int(input(f"Player {player}, enter position (1-9): ")) - 1
    board[move] = player

    wins = [(0,1,2), (3,4,5), (6,7,8), (0,3,6), (1,4,7), (2,5,8), (0,4,8), (2,4,6)]
    if any(board[a] == board[b] == board[c] for a, b, c in wins):
        print(board[0], board[1], board[2])
        print(board[3], board[4], board[5])
        print(board[6], board[7], board[8])
        print(f"Player {player} Wins!")
        break

    if player == "X":
        player = "O"
    else:
        player = "X"
else:
    print("Draw!")
```

```python
def vacuum_agent(location, state):
    if state[location] == 'DIRTY':
        state[location] = 'CLEAN'
        print(f"Cleaned Room {location}")
    else:
        location = 'B' if location == 'A' else 'A'
        print(f"Moved to Room {location}")
    return location

state = {'A': 'DIRTY', 'B': 'DIRTY'}
loc = 'A'

while 'DIRTY' in state.values():
    loc = vacuum_agent(loc, state)

print("Both rooms are CLEAN!")
```
