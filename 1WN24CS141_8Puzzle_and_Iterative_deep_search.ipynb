```python
def dfs(start):
    goal=(1,2,3,4,5,6,7,8,0)
    stack=[(start,[])]
    visited=set()
    print("KRISHA - 1WN24CS141")


    while stack:
        s,path=stack.pop()
        if s in visited: continue
        visited.add(s)

        if s==goal:
            print(path)
            return

        z=s.index(0)
        r,c=divmod(z,3)

        for dr,dc,m in [(-1,0,'U'),(1,0,'D'),(0,-1,'L'),(0,1,'R')]:
            nr,nc=r+dr,c+dc
            if 0<=nr<3 and 0<=nc<3:
                t=list(s); nz=nr*3+nc
                t[z],t[nz]=t[nz],t[z]
                stack.append((tuple(t),path+[m]))

    print("No solution")

start=tuple(map(int,input().split()))
dfs(start)
```
